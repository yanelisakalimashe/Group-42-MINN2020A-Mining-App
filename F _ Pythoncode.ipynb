```python

def monitor_equipment_maintenance():
    print("=== EQUIPMENT MAINTENANCE MONITORING ===")

    equipment_id = input("Enter equipment ID: ")

    try:
        operating_time = float(
            input("Enter operating time (hours): ")
        )
        downtime = float(
            input("Enter downtime (hours): ")
        )
        hours_since_service = float(
            input("Hours since last maintenance: ")
        )
        maintenance_interval = float(
            input("Maintenance interval (hours): ")
        )
        number_of_alerts = int(
            input("Number of equipment alerts: ")
        )

        if (
            operating_time < 0
            or downtime < 0
            or hours_since_service < 0
            or maintenance_interval <= 0
            or number_of_alerts < 0
        ):
            print("Error: Enter valid non-negative values.")
            print("Maintenance interval must be greater than zero.")
            return

    except ValueError:
        print("Error: Enter valid numerical values.")
        return

    maintenance_status = input(
        "Is the equipment under maintenance? (yes/no): "
    ).strip().lower()

    if maintenance_status not in ("yes", "no"):
        print("Error: Enter yes or no.")
        return

    # Calculate availability
    total_time = operating_time + downtime

    if total_time > 0:
        availability = (operating_time / total_time) * 100
    else:
        availability = 0

    print("\n--- MAINTENANCE REPORT ---")
    print("Equipment ID:", equipment_id)
    print(f"Operating time: {operating_time:.2f} hours")
    print(f"Downtime: {downtime:.2f} hours")
    print(f"Equipment availability: {availability:.2f}%")

    # Maintenance due and overdue checks
    if hours_since_service >= maintenance_interval:
        print("STATUS: Maintenance is due.")

        if ( hours_since_service > maintenance_interval and maintenance_status != "yes"):
            print("OVERDUE: Maintenance is required.")
    else:
        print("STATUS: Maintenance is not yet due.")

    # Equipment under maintenance
    if maintenance_status == "yes":
        print("STATUS: Equipment is under maintenance.")

    # Excessive downtime: prototype threshold of 20%
    if (total_time > 0 and downtime / total_time >= 0.20 ):
        print("WARNING: Excessive equipment downtime.")

    # Repeated alerts: prototype threshold of 3
    if number_of_alerts >= 3:
        print("WARNING: Repeated equipment alerts.")

    # Low availability: prototype threshold of 85%
    if availability < 85:
        print("WARNING: Low equipment availability.")

    print("\nMaintenance monitoring completed.")


# Run the program
monitor_equipment_maintenance()

```
