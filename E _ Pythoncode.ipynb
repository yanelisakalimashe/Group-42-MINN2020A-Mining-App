```python

def check_equipment_condition():
    # Input equipment information
    equipment_id = input("Enter equipment ID: ")

    try:
        temperature = float(input("Enter temperature (°C): "))
        vibration = float(input("Enter vibration (mm/s): "))

        # Validate sensor readings
        if temperature < 0 or vibration < 0:
            print("Error: Temperature and vibration cannot be negative.")
            return

    except ValueError:
        print("Error: Please enter valid numerical readings.")
        return

    print("\n--- EQUIPMENT CONDITION REPORT ---")
    print("Equipment ID:", equipment_id)

    # Temperature monitoring
    if temperature < 80:
        print("\nTemperature Status: NORMAL")
        print("Equipment operating within defined monitoring limits.")

    elif temperature <= 100:
        print("\nTemperature Status: WARNING")
        print("Equipment temperature approaching predefined threshold.")

    else:
        print("\nTemperature Status: CRITICAL")
        print("Equipment temperature exceeds predefined threshold.")
        print("Inspection required.")

    # Vibration monitoring
    if vibration < 5:
        print("\nVibration Status: NORMAL")
        print("Equipment operating within defined monitoring limits.")

    elif vibration <= 8:
        print("\nVibration Status: WARNING")
        print("Equipment vibration approaching predefined threshold.")

    else:
        print("\nVibration Status: CRITICAL")
        print("Equipment vibration exceeds predefined threshold.")
        print("Inspection required.")


# Run the program
check_equipment_condition()
```
