```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Import all sheets from the Excel workbook at once
df = pd.read_excel("C:/Users/FIHLANE/OneDrive/Desktop/MINN2020A_Complete_Realistic_Project_Data_CLEANED.xlsx",sheet_name=None)

# RISK ASSESSMENT
def risk_score():
    Likelihood = float(input("Enter the likelihood of the risk on a scale of (1-5)"))
    Consequence = float(input("Enter the consequences of the risk on a scale of (1-5)"))
    Risk_Score = Likelihood*Consequence
    if Risk_Score < 0 or Risk_Score>25:
        Classification = "Invalid"
        print("Invalid Risk Score")
    elif Risk_Score <=4:
        Classification = "Low"
    elif Risk_Score <=9:
        Classification = "Medium"
    elif Risk_Score <=16:
        Classification = "High"
    else:
        Classification = "Critical"
    print("Risk Score:",Risk_Score, "Classification:",Classification)
    # Warning for high or critical risk identified
    if Classification == "High":
        print("WARNING! High Risk identified; immediate attention required.")
    elif Classification == "Critical":
        print("WARNING! Critical Risk identified; immediate attention required.")


# Alerts and Notification
def alerts_monitoring():
    # Extract the required sheets from the imported data
    equipment = df["Equipment"]
    maintenance = df["Maintenance"]
    alerts = df["Alerts"]
    incidents = df["Safety Incidents"]
    observations = df["Safety Observations"]
    risks = df["Risk Assessments"]
    worker_safety = df["Worker Safety"]

    # Ask the user to enter an equipment ID
    equipment_id = input("\nEnter equipment ID: ").strip().upper()
    # Find the equipment in the imported data
    record = equipment[equipment["Equipment ID"] == equipment_id]
    # Check whether the equipment exists
    if record.empty:
        print("Equipment ID not found.")
        return
    # Get the equipment details
    row = record.iloc[0]
    print("\nMINING MONITORING ALERT")
    print("Equipment ID:", equipment_id)
    print("Equipment Type:", row["Equipment Type"])

    # Check temperature
    if row["Temperature (°C)"] > 100:
        print("CRITICAL: High temperature!")
    elif row["Temperature (°C)"] >= 80:
        print("WARNING: High temperature!")

    # Check vibration
    if row["Vibration (mm/s)"] > 8:
        print("CRITICAL: High vibration!")
    elif row["Vibration (mm/s)"] >= 5:
        print("WARNING: High vibration!")

    # Check equipment downtime
    if row["Downtime (h)"] > 24:
        print("WARNING: Excessive equipment downtime!")

    # Check maintenance status in the equipment sheet
    if str(row["Maintenance Status"]).lower() in ["due", "overdue"]:
        print("WARNING: Maintenance is due or overdue!")

    # Check overdue maintenance records for this equipment
    tasks = maintenance[(maintenance["Equipment ID"] == equipment_id) & (maintenance["Status"].str.lower() == "overdue")]
    for _, task in tasks.iterrows():
        print("OVERDUE MAINTENANCE:", task["Maintenance Type"])

    # Display existing alerts linked to this equipment
    equipment_alerts = alerts[(alerts["Record ID"] == equipment_id) & (alerts["Source"].str.lower() == "equipment")]
    for _, alert in equipment_alerts.iterrows():
        print(alert["Severity"], "ALERT:", alert["Parameter"],"-", alert["Recommended Action"])

    # Display critical safety incidents recorded in the mine
    critical = incidents[incidents["Severity"].str.lower() == "critical"]
    if not critical.empty:
        print("NOTICE:", len(critical),
              "critical safety incident(s) recorded in the mine.")

    # Display high-risk safety observations
    high_risk = observations[observations["Risk Level"].str.lower() == "high"]
    if not high_risk.empty:
        print("NOTICE:", len(high_risk),"high-risk safety observation(s) recorded in the mine.")

    # Display high or critical risk assessments
    high_risks = risks[risks["Risk Classification"].str.lower().isin(["high", "critical"])]
    if not high_risks.empty:
        print("NOTICE:", len(high_risks),
              "high/critical risk assessment(s) recorded.")
    print("Monitoring check complete.")
    
risk_score()
alerts_monitoring()
```
