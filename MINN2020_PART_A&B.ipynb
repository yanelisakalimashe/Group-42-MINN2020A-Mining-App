```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_excel("MINN2020A_Complete_Realistic_Project_Data.xlsx",sheet_name="Users")

print(df.columns.tolist())

permissions = {
    "Administrator": [
        "Dashboard", "Worker Safety", "Safety Incidents",
        "Equipment", "Maintenance", "Risk Assessment",
        "Reports", "Manage Users"
    ],

    "Safety Officer": [
        "Dashboard", "Worker Safety", "Safety Incidents",
        "Risk Assessment", "Reports"
    ],

    "Mining Engineer": [
        "Dashboard", "Worker Safety", "Safety Incidents",
        "Equipment", "Maintenance", "Risk Assessment",
        "Reports"
    ],

    "Maintenance Engineer": [
        "Dashboard", "Equipment", "Maintenance", "Reports"
    ],

    "Manager": [
        "Dashboard", "Worker Safety", "Safety Incidents",
        "Equipment", "Maintenance", "Risk Assessment",
        "Reports"
    ]
}

def dashboard():
    print("Dashboard for Mining Monitoring ")
    print("1. Worker Health and Safety")
    print("2. Safety Incidents")
    print("3. Equipment Monitoring")
    print("4. Maintenance Monitoring")


print(df.head())
print(df.columns)

username=input("Enter Username:")
password=input("Enter Password:")

user=df[df["Username"]==username]

if user.empty:
    print("Username not found")
else:
    print("username found")

stored_pword = user["Password"].iloc[0]
if password == stored_pword:
    print("Login successful")

    role = user["Role"].iloc[0]
    print("Welcome", username, "your role is:", role)

    if role in permissions:
        allowed_functions = permissions[role]
        print("You can access:")

        for function in allowed_functions:
            print("-", function)

        choice = input("Enter the function you want to access: ").strip()

        if choice in allowed_functions:
            print("Access granted to", choice)

            if choice == "Dashboard":
                dashboard()

            if choice == "Worker Safety":
                worker_safety()

            else:
                print("This function is still being developed")
        
        else:
            print("Access denied")


    else:
        print("Unknown role")
        
    

    if role == "Admnistrator":
        print("Full access granted")
    elif role == "Safety Officer":
        print("Access to Safety Officer functions granted")
    elif role == "Mining Engineer":
        print("Access to Mining Engineer functions granted")
    elif role == "Maintenance Engineer":
        print("Access to Maintenance Engineer functions granted")
    elif role == "Manager":
        print("Access to management functions granted")
    else:
        print("Unknown role")
    
else:
    print("Incorrect password")


# part B
def worker_safety():
    print("WORKER HEALTH AND SAFETY MONITORING")

    safety_df = pd.read_excel("MINN2020A_Complete_Realistic_Project_Data.xlsx", sheet_name="Worker Safety")
    print(safety_df)

    print("TEST: Safety warning section reached")

    print("SAFETY WARNINGS!!!!")

    for index, worker in safety_df.iterrows():
        
        if worker["PPE Compliance"] == "No":
            print("WARNING:", worker["Worker ID"], "- PPE not compliant")

        if worker["Safety Training Status"] == "Incomplete":
            print("WARNING:", worker["Worker ID"], "- Training incomplete")

        if worker["Fatigue Level"] == "High":
            print("WARNING:", worker["Worker ID"], "- High fatigue")

        if worker["Risk level"] == "High":
            print("WARNING:", worker["Worker ID"], "- High safety risk")

        elif worker["Risk level"] == "Critical":
            print("CRITICAL:", worker["Worker ID"], "- Immediate attention required") 
            
            
def add_worker():

    print("ADD NEW WORKER SAFETY RECORD")

    filename = "MINN2020A_Complete_Realistic_Project_Data.xlsx"

    safety_df = pd.read_excel(filename, sheet_name="Worker Safety")

    worker_id = input("Enter Worker ID: ").strip().upper()

    if worker_id in safety_df["Worker ID"].astype(str).values:
        print("ERROR: Worker ID already exists")
        return

    if worker_id == "":
        print("ERROR: Worker ID cannot be empty")
        return

    department = input("Enter Department: ").strip()
    ppe = input("PPE Compliance (Yes/No): ").strip().title()
    training = input("Safety Training (Completed/Incomplete): ").strip().title()
    fatigue = input("Fatigue Level (Low/Medium/High): ").strip().title()
    risk = input("Risk Level (Low/Medium/High/Critical): ").strip().title()

    if department == "":
        print("ERROR: Department cannot be empty")
        return

    if ppe not in ["Yes", "No"]:
        print("ERROR: PPE must be Yes or No")
        return

    if training not in ["Completed", "Incomplete"]:
        print("ERROR: Invalid training status")
        return

    if fatigue not in ["Low", "Medium", "High"]:
        print("ERROR: Invalid fatigue level")
        return

    if risk not in ["Low", "Medium", "High", "Critical"]:
        print("ERROR: Invalid risk level")
        return

    new_worker = pd.DataFrame({"Worker ID": [worker_id],
        "Department": [department],
        "PPE Compliance": [ppe],
        "Safety Training Status": [training],
        "Fatigue Level": [fatigue],
        "Risk level": [risk]
    })

    safety_df = pd.concat([safety_df, new_worker], ignore_index=True)

    with pd.ExcelWriter(filename, engine="openpyxl", mode="a",
                        if_sheet_exists="replace") as writer:
        safety_df.to_excel(writer, sheet_name="Worker Safety", index=False)

    print("Worker record saved successfully!")
    print("Worker ID:", worker_id)

add_worker()
```
